## 1 · Habla con el notebook

**Qué es.** Python es un lenguaje de programación: escribes instrucciones con reglas precisas y el computador las ejecuta en orden, de arriba hacia abajo. Un **notebook** es un documento con dos tipos de celdas:

| Celda | Qué contiene | Qué ocurre al ejecutarla |
|---|---|---|
| Texto (Markdown) | Explicaciones como esta | Se muestra con formato |
| Código | Instrucciones de Python | Python las ejecuta y muestra el resultado debajo |

**Escribir no es ejecutar.** Puedes escribir código en una celda, pero Python no hace nada hasta que la ejecutas.

**Tu primera instrucción.** `print(...)` es una **función**: un nombre seguido de paréntesis. Lo que va dentro de los paréntesis es el **argumento**, el dato que la función recibe. `print` lo muestra en pantalla.

````python
# Una línea que comienza con # es un comentario: Python no la ejecuta.
print("Torre de Bogotá: recibido")
print(2 + 3)
````

**Lee lo que pasó, línea por línea:**

1. `# ...` es un **comentario**: una nota para personas. Python lo ignora; por eso no produce resultado.
2. `print("Torre de Bogotá: recibido")` muestra el **texto** que está entre comillas, tal cual.
3. `print(2 + 3)` no tiene comillas: Python primero **calcula** `2 + 3` y después muestra `5`.

- El **documento** conserva texto, código y los resultados que se mostraron.
- El **kernel** conserva los datos **solo mientras está activo**. Si lo reinicias (en Colab: **Entorno de ejecución → Reiniciar sesión**), esa memoria se vacía, aunque en pantalla sigas viendo resultados antiguos.

## 2 · Una calculadora que recuerda: variables

**Qué es.** Una **variable** es un nombre que apunta a un valor guardado en la memoria del kernel. Sirve para no repetir números y para que cada cálculo diga *qué* representa.

**Cómo se escribe.** `distancia_km = 420` se lee «guarda 420 con el nombre `distancia_km`». Python evalúa primero lo que está a la **derecha** del `=` y después lo guarda con el nombre de la **izquierda**. Por eso `=` es **asignación**, no igualdad matemática; para comparar usaremos `==` en la sección 4.

Reglas útiles para nombres:

- Sin espacios: usa `_` (`tiempo_h`, no `tiempo h`). Un nombre no puede empezar con un número.
- Mayúsculas y minúsculas importan: `tiempo_h` y `Tiempo_h` son variables distintas.
- Incluye la unidad en el nombre (`_km`, `_h`, `_w`): Python no conoce unidades; tú sí.
- El separador decimal es el punto: escribe `1.5`, no `1,5`.

| Escritura | Operación | Ejemplo | Resultado |
|---|---|---|---|
| `+`, `-` | Sumar, restar | `4 - 1` | `3` |
| `*`, `/` | Multiplicar, dividir | `6 / 2` | `3.0` |
| `**` | Potencia | `2 ** 3` | `8` |
| `( )` | Agrupar | `(2 + 3) * 4` | `20` |

Cuidado: `^` **no** es potencia en Python. Usa `**`.

Estas cifras describen un simulador, no un plan de vuelo. **Predice** el tiempo antes de ejecutar.

````python
distancia_km = 420
velocidad_kmh = 140
tiempo_h = distancia_km / velocidad_kmh
print(f"Tiempo: {tiempo_h:.1f} h")
````

**Qué ocurrió.** Las dos primeras líneas guardan datos. La tercera **usa** esos nombres: Python busca sus valores en memoria, divide y guarda el resultado en `tiempo_h`. La cuarta lo muestra.

**f-strings: texto con valores dentro.** Un texto con `f` antes de las comillas (`f"..."`) puede incluir variables entre llaves `{ }`. Después de `:` indicas el formato:

| Escribes | Muestra | Significado |
|---|---|---|
| `{tiempo_h}` | `3.0` | El valor tal cual |
| `{tiempo_h:.1f}` | `3.0` | Un decimal |
| `{tiempo_h:.3f}` | `3.000` | Tres decimales |
| `{0.000275:.3g}` | `0.000275` | Tres cifras significativas |

El formato cambia **cómo se ve** el número, no el valor guardado.

**Unidades.** La fórmula conserva las unidades: km ÷ (km/h) = h. Python no verifica unidades por ti; los nombres con unidad te ayudan a detectar errores.

### Cambiar un valor y escribir números grandes

Una variable puede recibir un valor nuevo. La celda siguiente muestra tres ideas que usarás constantemente. **Predice** las cuatro líneas de salida.

````python
# 1. Reasignar: el nuevo valor reemplaza al anterior.
velocidad_kmh = 210
print(f"Con {velocidad_kmh} km/h: {distancia_km / velocidad_kmh:.1f} h")

# 2. Usar el valor actual para calcular el siguiente.
vueltas = 0
vueltas = vueltas + 1
print("Vueltas:", vueltas)

# 3. Notación científica: 3e8 significa 3 × 10**8.
velocidad_luz_m_s = 3e8
print(f"Luz: {velocidad_luz_m_s:.0f} m/s")
print(f"Un microsegundo: {1e-6} s")
````

**Observa:**

- `vueltas = vueltas + 1` es imposible como ecuación, pero tiene sentido como instrucción: «toma el valor actual, súmale 1 y guárdalo con el mismo nombre».
- `3e8` es notación científica (3 × 10⁸) y `1e-6` vale 0.000001. En el Laboratorio 02 escribirás así frecuencias (`1090e6` Hz), tiempos (`1e-6` s) y potencias.

## 3 · Valores de distintas clases: tipos

## 4 · Varios datos y una decisión

### Listas

**Qué es.** Una **lista** guarda varios valores en orden, entre corchetes y separados por comas. Sirve cuando tienes una serie de medidas: duraciones, distancias, muestras de una señal.

- **Índice:** cada elemento tiene una posición que **empieza en 0**. `duraciones_min[0]` es el primero; `duraciones_min[-1]`, el último.
- **Porciones:** `duraciones_min[0:2]` toma desde la posición 0 **hasta antes** de la 2, es decir, los dos primeros. `duraciones_min[:2]` significa lo mismo.
- `len()` cuenta elementos; `sum()` los suma.

No mezcles unidades en una misma lista de medidas. **Predice** las cinco líneas de salida.

````python
duraciones_min = [20, 35, 25]
print("Primera etapa:", duraciones_min[0])
print("Última etapa:", duraciones_min[-1])
print("Dos primeras:", duraciones_min[:2])
print("Etapas:", len(duraciones_min))
print("Total:", sum(duraciones_min))
````

`print("Total:", sum(duraciones_min))` muestra varios argumentos separados por comas; `print` los une con un espacio. Es una alternativa sencilla a las f-strings.

### Comparaciones y decisiones

Una **comparación** produce un `bool`: `True` o `False`.

| Escritura | Pregunta |
|---|---|
| `a == b` | ¿Son iguales? |
| `a != b` | ¿Son distintos? |
| `a > b`, `a < b` | ¿Mayor?, ¿menor? |
| `a >= b`, `a <= b` | ¿Mayor o igual?, ¿menor o igual? |

Un **`if`** decide qué bloque ejecutar:

- La línea del `if` termina con **dos puntos** `:`.
- Las líneas del bloque llevan **sangría** de cuatro espacios. La sangría no es decoración: indica qué instrucciones pertenecen al bloque.
- `else:` agrupa lo que ocurre cuando la condición es `False`.

En el Laboratorio 02 usarás esta estructura para decidir si una señal supera un **umbral** de detección.

**Predice** qué mensaje aparecerá con un total de 80 min.

## 5 · Una fórmula reutilizable: funciones

**Por qué.** Si vas a repetir el mismo cálculo con datos distintos, conviene escribirlo **una vez** y darle un nombre. Eso es una **función**. Ya usaste funciones de Python (`print`, `len`, `sum`); ahora crearás las tuyas.

```python
def minutos_a_horas(minutos):    # def + nombre + (parámetro) + dos puntos
    return minutos / 60          # bloque con sangría; return entrega el resultado
```

- **Parámetro:** `minutos` es el nombre que recibe el dato de entrada **dentro** de la función.
- **Argumento:** el valor concreto que pasas al **llamarla**, por ejemplo `minutos_a_horas(90)`.
- **`return`:** entrega el resultado a quien llamó la función, para guardarlo o usarlo en otro cálculo. `print` solo muestra; no entrega nada.

**Definir no es ejecutar.** La parte con `def` solo enseña a Python una receta; el cálculo ocurre cuando la **llamas**.

**Predice** las dos líneas antes de ejecutar.

````python
def minutos_a_horas(minutos):
    return minutos / 60

print(f"Una hora y media: {minutos_a_horas(90):.1f} h")
print(f"Dos horas: {minutos_a_horas(120):.1f} h")
````

### Varios parámetros y valores por defecto

Una función puede recibir varios datos. Un parámetro puede tener un **valor por defecto**, que se usa si no lo indicas al llamarla. También puedes **nombrar** los argumentos (`velocidad_kmh=280`): el código explica qué significa cada número.

En el Laboratorio 02 llamarás funciones así: `radar_w(60e3, pico_kw=100)` calcula la potencia de un eco cambiando solo la potencia transmitida.

**Predice** las tres líneas antes de ejecutar.

````python
def tiempo_vuelo_h(distancia_km, velocidad_kmh=140):
    """Devuelve el tiempo, en horas, para recorrer una distancia a velocidad constante."""
    return distancia_km / velocidad_kmh

print(f"Por defecto: {tiempo_vuelo_h(420):.1f} h")
print(f"Más rápido:  {tiempo_vuelo_h(420, velocidad_kmh=280):.1f} h")
dos_tramos_h = tiempo_vuelo_h(210) + tiempo_vuelo_h(210)
print(f"Dos tramos:  {dos_tramos_h:.1f} h")
````

**Observa:** el texto entre comillas triples bajo el `def` es un **docstring**: una descripción de la función para quien la use. En la tercera llamada, `return` permite **sumar** los resultados de dos llamadas; con un `print` dentro de la función eso no sería posible.

## 6 · Repetir sin copiar: el ciclo `for`

**Por qué.** A menudo queremos aplicar el mismo cálculo a varios valores: varias distancias, varias frecuencias, varios escenarios. Copiar la misma línea diez veces es lento y propenso a errores.

**Cómo se escribe.** `for velocidad in velocidades_kmh:` significa «toma cada elemento de la lista, uno por uno, llámalo `velocidad` y ejecuta el bloque con sangría». En cada vuelta, `velocidad` vale el elemento siguiente.

Calcular lo mismo para una serie de valores se llama **barrido de parámetros**; en el Laboratorio 02 barrerás distancias, frecuencias y ángulos.

**Predice:** ¿cuántas líneas imprimirá la celda?

````python
velocidades_kmh = [140, 210, 280]
for velocidad in velocidades_kmh:
    duracion_h = tiempo_vuelo_h(420, velocidad_kmh=velocidad)
    print(f"{velocidad} km/h → {duracion_h:.1f} h")

# Forma compacta: una lista nueva con un cálculo por elemento.
tiempos_h = [tiempo_vuelo_h(420, velocidad_kmh=v) for v in velocidades_kmh]
print("Lista de tiempos:", tiempos_h)
````

**Lee la última parte.** `[expresión for v in lista]` es una **comprensión de lista**: crea una lista nueva con el resultado de la expresión para cada elemento. Hace lo mismo que el ciclo, en una sola línea. La verás alguna vez en el Laboratorio 02; si te resulta difícil de leer, piensa en el `for` equivalente.

## 7 · Una caja de herramientas: librerías y NumPy

**Qué es una librería.** Código que otras personas ya escribieron y probaron. `import` la carga en la sesión para que puedas usarla:

- **math** viene con Python: funciones matemáticas para números individuales (`math.sqrt`, `math.pi`).
- **NumPy** calcula con **conjuntos** de números de una sola vez. Es la herramienta numérica principal del Laboratorio 02.
- **Matplotlib** dibuja gráficos.

`import numpy as np` significa «carga NumPy y llámalo `np`». Después escribes `np.mean(...)`: el punto indica que `mean` pertenece a NumPy. `as np` y `as plt` son abreviaturas que usa casi todo el mundo.

En Colab estas librerías ya están instaladas: **importar no es instalar**. Si reinicias el kernel, debes volver a ejecutar los `import`.

````python
import math
import numpy as np
import matplotlib.pyplot as plt

print(f"Raíz de 9: {math.sqrt(9):.1f}")
duraciones = np.array([20, 35, 25])
print("Cada etapa en horas:", np.round(duraciones / 60, 2))
print(f"Promedio: {np.mean(duraciones):.2f} min")
````

**Qué hizo `np.array`.** Un **array** de NumPy se parece a una lista, pero permite operar con todos sus números de una vez: `duraciones / 60` divide cada elemento. Con una lista normal, `[20, 35, 25] / 60` produce un `TypeError`.

`np.round(..., 2)` redondea **lo que se muestra**; conserva los valores originales para seguir calculando. `np.mean` calcula el promedio.

### Herramientas de NumPy que usarás en el Laboratorio 02

| Herramienta | Qué produce | Uso posterior |
|---|---|---|
| `np.linspace(a, b, n)` | `n` valores equiespaciados de `a` a `b`, ambos incluidos | Distancias de 10 a 200 km para dibujar una curva |
| `np.arange(n)` | Enteros `0, 1, …, n-1` | Número de cada muestra de una señal |
| `np.log10(x)` | Logaritmo en base 10: «¿10 elevado a qué da x?» | Convertir potencias a decibelios |
| `np.pi`, `np.sqrt`, `np.sin` | Constante π, raíz cuadrada, seno | Ondas y antenas |

Una **comparación sobre un array** produce un array de `True`/`False`, uno por elemento: así se aplica un umbral a muchas medidas a la vez.

**Predice** la salida de cada línea antes de ejecutar.

````python
distancias_km = np.linspace(0, 100, 5)
print("linspace:", distancias_km)
print("arange:", np.arange(5))
print("log10 de 1000:", np.log10(1000))
print("10 elevado a 3:", 10 ** 3)

medidas = np.array([12, 31, 8, 27])
supera = medidas > 25
print("¿Supera 25?:", supera)
print("Cuántas superan:", np.count_nonzero(supera))
````

**Interpreta.** `np.log10(1000)` vale 3 porque 10³ = 1000: el logaritmo «cuenta ceros». Por eso los decibelios expresan números enormes y diminutos con cifras manejables. `[False  True False  True]` marca qué medidas superan el umbral; `np.count_nonzero` cuenta los `True`.

## 8 · Ver los números: gráficos con Matplotlib

Un gráfico convierte una tabla de números en una forma que el ojo compara rápido. Todo gráfico útil responde a tres preguntas: **¿qué se mide?, ¿en qué unidades?, ¿qué relación muestra?**

**Predice:** ¿qué etapa tendrá la barra más alta?

````python
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(["Preparación", "Práctica", "Cierre"], duraciones, color="#1672BC")
ax.set_ylabel("Duración [min]")
ax.set_title("Registro de una sesión de simulador")
ax.grid(axis="y", alpha=0.2)
plt.show()
````

**Lee el código, línea por línea:**

1. `fig, ax = plt.subplots(figsize=(7, 3))` crea una figura de 7 × 3 pulgadas. Devuelve **dos** objetos y los guarda en dos variables a la vez: `fig`, la figura completa, y `ax`, los ejes, el área donde se dibuja.
2. `ax.bar(categorías, valores, color=...)` dibuja barras.
3. `ax.set_ylabel` y `ax.set_title` escriben la etiqueta del eje y el título. Sin unidades, un gráfico no se puede interpretar.
4. `ax.grid(...)` añade una cuadrícula tenue; `plt.show()` muestra el resultado.

**Interpreta:** la barra más alta es la etapa más larga. **Modifica:** cambia un valor de `duraciones` en la celda de NumPy, ejecútala y vuelve a ejecutar esta. ¿Cambió el gráfico?

### Una relación entre dos magnitudes

En el Laboratorio 02 casi todos los gráficos son **curvas**: muestran cómo cambia una magnitud cuando otra varía.

- `ax.plot(x, y)` une los puntos `(x, y)` con una línea.
- `label=` da nombre a cada curva y `ax.legend()` muestra esos nombres.
- `ax.set(xlabel=..., ylabel=..., title=...)` fija varias etiquetas en una sola llamada.

**Predice:** ¿qué recta subirá más rápido?

````python
horas_vuelo = np.linspace(0, 3, 31)
fig, ax = plt.subplots(figsize=(7, 3))
for velocidad in [140, 280]:
    ax.plot(horas_vuelo, velocidad * horas_vuelo, label=f"{velocidad} km/h")
ax.set(xlabel="Tiempo [h]", ylabel="Distancia recorrida [km]",
       title="A mayor velocidad, mayor pendiente")
ax.grid(alpha=0.2)
ax.legend()
plt.show()
````

**Interpreta.** Ambas rectas parten de 0 km. La de 280 km/h tiene el doble de pendiente: en el mismo tiempo recorre el doble de distancia. La **forma** de una curva revela la relación física. En el Laboratorio 02 verás curvas de potencia que caen con la distancia y tendrás que explicar por qué una cae más rápido que otra.

## Si algo falla

Un mensaje de error de Python (*traceback*) puede ser largo. Empieza por la **última línea**: dice el tipo de error y una pista. Las líneas anteriores indican dónde ocurrió.

| Mensaje | Qué suele significar | Revisa |
|---|---|---|
| `NameError` | Python no conoce ese nombre | Ortografía, mayúsculas y si ejecutaste la celda donde se creó (por ejemplo, después de reiniciar) |
| `SyntaxError` | La instrucción está mal escrita | Comillas, paréntesis y dos puntos |
| `IndentationError` | La sangría no es coherente | Espacios al inicio del bloque |
| `TypeError` | Operación entre tipos incompatibles | Texto, número, lista o array: ¿son compatibles? |
| `ZeroDivisionError` | División entre cero | El valor del divisor |
| Sin error, pero el resultado no cuadra | Celdas ejecutadas fuera de orden o variable reemplazada | Reinicia y ejecuta todo |
